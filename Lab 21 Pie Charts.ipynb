### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
# Drop missing values from our target database interest tracking column
df_db = df['DatabaseWantToWorkWith'].dropna()

# Split semicolon strings into lists and explode them into distinct individual rows
df_exploded = df_db.str.split(';').explode()

# Grab the counts of the top 5 most-desired databases
top_5_databases = df_exploded.value_counts().nlargest(5)

# Initialize graphic canvas size layout
plt.figure(figsize=(9, 9))

# Plot the pie chart configuration matrix
plt.pie(
    top_5_databases, 
    labels=top_5_databases.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#2b5c8f', '#4682b4', '#1f77b4', '#aec7e8', '#ff7f0e'],
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5}
)

# Style text properties and formatting bounds
plt.title('Top 5 Most Desired Databases Respondents Want to Work With', fontsize=14, pad=20)

# Ensure the pie is drawn as a perfect circle
plt.axis('equal') 

# Display visual graph dashboard layout
plt.tight_layout()
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
# Drop missing values from our target Developer Type tracking column
df_dev = df['DevType'].dropna()

# Split the semicolon-separated strings into lists and explode them into distinct rows
df_exploded = df_dev.str.split(';').explode()

# Extract the frequency counts of the top 5 most represented developer roles
top_5_roles = df_exploded.value_counts().nlargest(5)

# Initialize layout canvas scale properties
plt.figure(figsize=(10, 10))

# Plot the pie chart configuration matrix
plt.pie(
    top_5_roles, 
    labels=top_5_roles.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#008080', '#20b2aa', '#4682b4', '#40e0d0', '#00ced1'], # Deep teal/blue tech theme
    shadow=True,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5, 'antialiased': True}
)

# Title formatting and layout constraints
plt.title('Top 5 Respondent Developer Roles Distribution Matrix', fontsize=14, pad=20)

# Ensure the pie is rendered as a perfect uniform circle
plt.axis('equal') 

# Display visual graph dashboard layout
plt.tight_layout()
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
# Drop missing values from our target professional operating system column
df_opsys = df['OpSysProfessional use'].dropna()

# Split the semicolon-separated strings into lists and explode them into distinct individual rows
df_exploded = df_opsys.str.split(';').explode()

# Extract the frequency counts of all professionally used operating systems
opsys_distribution = df_exploded.value_counts()

# Initialize graphic canvas size layout
plt.figure(figsize=(10, 10))

# Plot the pie chart configuration matrix
plt.pie(
    opsys_distribution, 
    labels=opsys_distribution.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#2b5c8f', '#4682b4', '#1f77b4', '#aec7e8', '#ff7f0e', '#d9534f', '#2ca02c'],
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5}
)

# Style text properties and formatting bounds
plt.title('Operating Systems Used by Respondents for Professional Use', fontsize=14, pad=20)

# Ensure the pie is drawn as a perfect circle on the canvas
plt.axis('equal') 

# Display visual graph dashboard layout
plt.tight_layout()
plt.show()
```
