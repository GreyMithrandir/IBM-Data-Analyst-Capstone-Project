#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
# --- 1. HANDLE EMPLOYMENT (Impute with Mode) ---
# Find the single most frequent employment type
employment_mode = df['Employment'].value_counts().idxmax()
df['Employment'].fillna(employment_mode, inplace=True)
print(f"Imputed 'Employment' with mode: '{employment_mode}'")


# --- 2. HANDLE REMOTEWORK (Impute with Mode) ---
# Find the single most frequent work environment type
remotework_mode = df['RemoteWork'].value_counts().idxmax()
df['RemoteWork'].fillna(remotework_mode, inplace=True)
print(f"Imputed 'RemoteWork' with mode: '{remotework_mode}'")


# --- 3. HANDLE JOBSAT (Drop or Flag) ---
# Strategy A: Drop rows missing JobSat to keep sentiment data pure (Standard Data Science)
df.dropna(subset=['JobSat'], inplace=True)
print("Dropped rows where 'JobSat' was missing.")

# Strategy B (Alternative): If you are forbidden from dropping rows in this lab, 
# fill it with a neutral or 'Missing' string placeholder instead:
# df['JobSat'].fillna('Neutral', inplace=True)
```

```python
# Check remaining null values for these specific columns
print("\nRemaining Missing Values:")
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())

# Check the new dataframe size
print("\nFinal shape of DataFrame:", df.shape)
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# Clean YearsCodePro (handles text edge cases from the survey)
df['YearsCodePro'] = df['YearsCodePro'].replace('Less than 1 year', 0)
df['YearsCodePro'] = df['YearsCodePro'].replace('More than 50 years', 51)
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df.dropna(subset=['YearsCodePro', 'JobSat'], inplace=True)

# Map JobSat strings to a numeric scale if not already numeric
# Higher values indicate higher satisfaction (1 = Lowest, 5 = Highest)
jobsat_mapping = {
    'Very dissatisfied': 1,
    'Slightly dissatisfied': 2,
    'Neither satisfied nor dissatisfied': 3,
    'Slightly satisfied': 4,
    'Very satisfied': 5
}
# Apply map only if columns contain strings
if df['JobSat'].dtype == 'object':
    df['JobSat_Numeric'] = df['JobSat'].map(jobsat_mapping)
else:
    df['JobSat_Numeric'] = df['JobSat']

# Step 1: Create the experience ranges using bin edges
bins = [-1, 5, 10, 20, np.inf]
labels = ['0-5 years', '5-10 years', '10-20 years', '>20 years']
df['ExperienceRange'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels)

# Step 2: Calculate median satisfaction for each group
satisfaction_summary = df.groupby('ExperienceRange')['JobSat_Numeric'].median().reset_index()
print(satisfaction_summary)
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='ExperienceRange', y='JobSat_Numeric', data=satisfaction_summary, palette='Blues_d')

plt.title('Median Job Satisfaction by Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience (YearsCodePro)')
plt.ylabel('Median Job Satisfaction Score (1-5)')
plt.ylim(0, 5.5) # Provide room at the top for labels
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
# Ensure all missing rows are dropped for an accurate count distribution
df_clean_sat = df.dropna(subset=['JobSat'])

# Define a strict ordering so the categories flow from least satisfied to most satisfied
jobsat_order = [
    'Very dissatisfied', 
    'Slightly dissatisfied', 
    'Neither satisfied nor dissatisfied', 
    'Slightly satisfied', 
    'Very satisfied'
]

# Set the visualization styling
plt.figure(figsize=(10, 6))
sns.countplot(
    y='JobSat', 
    data=df_clean_sat, 
    order=jobsat_order,
    palette='RdYlGn'  # Red-to-Green color palette matching satisfaction levels
)

# Customize titles and labels
plt.title('Overall Distribution of Respondent Job Satisfaction', fontsize=14, pad=15)
plt.xlabel('Number of Respondents (Count)', fontsize=12)
plt.ylabel('Job Satisfaction Level', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)

# Render the plot
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
# Ensure no missing values interfere with the plot
df_remote_clean = df.dropna(subset=['RemoteWork'])

plt.figure(figsize=(8, 5))
sns.countplot(
    x='RemoteWork', 
    data=df_remote_clean, 
    palette='viridis',
    order=df_remote_clean['RemoteWork'].value_counts().index
)

plt.title('Distribution of Remote Work Preferences')
plt.xlabel('Work Type Arrangement')
plt.ylabel('Number of Respondents')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Cross-tabulate with row-wise percentages (normalize='index')
employment_remote_cross = pd.crosstab(
    df['Employment'], 
    df['RemoteWork'], 
    normalize='index'
) * 100

print("Remote Work Preferences by Employment Type (Percentages):")
print(employment_remote_cross.round(1))
```

```python
# 1. Explode multi-select DevType values so each job role gets its own row
df_exploded = df.dropna(subset=['DevType', 'RemoteWork']).copy()
df_exploded['DevType'] = df_exploded['DevType'].str.split(';')
df_exploded = df_exploded.explode('DevType')

# 2. Cross-tabulate exploded job roles against remote work preferences
role_remote_cross = pd.crosstab(
    df_exploded['DevType'], 
    df_exploded['RemoteWork'], 
    normalize='index'
) * 100

# 3. Sort by 'Remote' descending to see which roles work remotely the most
role_remote_cross = role_remote_cross.sort_values(by='Remote', ascending=False)

print("Top 10 Roles with the Highest Percentage of Remote Work:")
print(role_remote_cross.head(10).round(1))
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# Basic regional mapping dictionary based on common Stack Overflow country strings
region_map = {
    'United States': 'North America', 'Canada': 'North America', 'Mexico': 'North America',
    'United Kingdom': 'Europe', 'Germany': 'Europe', 'France': 'Europe', 'Netherlands': 'Europe',
    'India': 'Asia', 'China': 'Asia', 'Japan': 'Asia', 'Singapore': 'Asia',
    'Brazil': 'South America', 'Argentina': 'South America',
    'Australia': 'Oceania', 'New Zealand': 'Oceania',
    'South Africa': 'Africa', 'Nigeria': 'Africa', 'Kenya': 'Africa'
}

# Apply the region mapping to a new column (drop unmapped countries for clean plotting)
df['Region'] = df['Country'].map(region_map)
df_regional = df.dropna(subset=['Region', 'LanguageHaveWorkedWith']).copy()

# Explode the semi-colon separated languages so each language gets its own row
df_regional['LanguageHaveWorkedWith'] = df_regional['LanguageHaveWorkedWith'].str.split(';')
df_exploded = df_regional.explode('LanguageHaveWorkedWith')

# Calculate the raw counts of language usage within each region
lang_region_counts = pd.crosstab(
    df_exploded['LanguageHaveWorkedWith'], 
    df_exploded['Region']
)

# Identify the top 10 global languages to keep the visualization clean
top_10_global_langs = df_exploded['LanguageHaveWorkedWith'].value_counts().head(10).index
lang_region_final = lang_region_counts.loc[top_10_global_langs]

print("Top Programming Languages by Region (Counts):")
print(lang_region_final)
```

```python
plt.figure(figsize=(12, 8))

# Normalize by column (Region) to show relative language popularity percentages within that region
sns.heatmap(
    lang_region_final.div(lang_region_final.sum(axis=0), axis=1) * 100, 
    annot=True, 
    cmap='YlGnBu', 
    fmt=".1f"
)

plt.title('Relative Popularity (%) of Top Programming Languages by Region', fontsize=14, pad=15)
plt.xlabel('Global Region', fontsize=12)
plt.ylabel('Programming Language', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Reset index to prepare for melting
lang_melted = lang_region_final.reset_index().melt(
    id_vars='LanguageHaveWorkedWith', 
    var_name='Region', 
    value_name='Count'
)

# Plot using a faceted or grouped bar layout
plt.figure(figsize=(14, 7))
sns.barplot(
    x='LanguageHaveWorkedWith', 
    y='Count', 
    hue='Region', 
    data=lang_melted, 
    palette='muted'
)

plt.title('Language Usage Frequencies Across Different Regions', fontsize=14)
plt.xlabel('Programming Language', fontsize=12)
plt.ylabel('Number of Respondents', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Regions')
plt.tight_layout()
plt.show()
```

```python
# 1. Clean and prepare data by dropping rows missing core columns
df_lang = df.dropna(subset=['LanguageWorkedWith', 'Region']).copy()

# 2. Explode the semicolon-separated languages into individual records
df_lang['LanguageWorkedWith'] = df_lang['LanguageWorkedWith'].str.split(';')
df_exploded = df_lang.explode('LanguageWorkedWith')

# 3. Identify the top 10 programming languages globally
top_10_langs = df_exploded['LanguageWorkedWith'].value_counts().head(10).index.tolist()

# 4. Filter the exploded dataframe down to only these top 10 languages
df_filtered = df_exploded[df_exploded['LanguageWorkedWith'].isin(top_10_langs)]

# 5. Cross-tabulate raw respondent counts (Languages vs Regions)
lang_region_counts = pd.crosstab(df_filtered['LanguageWorkedWith'], df_filtered['Region'])

# Reorder rows to match the global popularity ranking
lang_region_counts = lang_region_counts.loc[top_10_langs]

print(lang_region_counts)
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
# 1. Clean up YearsCodePro (handling survey text edge cases)
df['YearsCodePro'] = df['YearsCodePro'].replace('Less than 1 year', 0)
df['YearsCodePro'] = df['YearsCodePro'].replace('More than 50 years', 51)
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# 2. Drop rows missing either critical variable
df_scatter_clean = df.dropna(subset=['YearsCodePro', 'JobSatPoints_1']).copy()

# 3. Ensure satisfaction score is numeric
df_scatter_clean['JobSatPoints_1'] = pd.to_numeric(df_scatter_clean['JobSatPoints_1'], errors='coerce')

# 4. Generate the Scatter Plot
plt.figure(figsize=(10, 6))

# We use regplot to add a linear regression line to visually track the overall upward/downward trend.
# x_jitter introduces a tiny offset so you can see the dense clusters of individual data points.
sns.regplot(
    x='YearsCodePro', 
    y='JobSatPoints_1', 
    data=df_scatter_clean,
    scatter_kws={'alpha':0.3, 'color': '#2b7bba'}, # Alpha transparency helps reveal point density
    line_kws={'color': '#e65c00', 'linewidth': 2},  # Bright trend line
    x_jitter=0.2, 
    y_jitter=0.1
)

# Customize title and axis configurations
plt.title('Correlation Between Professional Coding Experience and Job Satisfaction Score', fontsize=14, pad=15)
plt.xlabel('Years of Professional Coding Experience (YearsCodePro)', fontsize=12)
plt.ylabel('Job Satisfaction Score (JobSatPoints_1)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Render the plot
plt.tight_layout()
plt.show()

# 5. Calculate and print the correlation coefficient
correlation = df_scatter_clean['YearsCodePro'].corr(df_scatter_clean['JobSatPoints_1'])
print(f"Pearson Correlation Coefficient: {correlation:.4f}")
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
# Clean up EdLevel strings into concise, readable categories
ed_mapping = {
    'Bachelor’s degree (B.A., B.S., B.Eng., etc.)': "Bachelor's",
    'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)': "Master's",
    'Other doctoral degree (Ph.D., Ed.D., etc.)': 'Ph.D./Doctoral',
    'Primary/elementary school': 'Primary School',
    'Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)': 'High School',
    'Some college/university study without earning a degree': 'Some College (No Degree)',
    'Associate degree (A.A., A.S., etc.)': "Associate's",
    'Professional degree (JD, MD, etc.)': 'Professional Degree',
    'I never completed any formal education': 'No Formal Education'
}
df['EdLevel_Clean'] = df['EdLevel'].replace(ed_mapping)

# Drop missing values in the target columns
df_cross = df.dropna(subset=['EdLevel_Clean', 'Employment']).copy()

# Cross-tabulate and normalize by row index to get percentages
ed_emp_crosstab = pd.crosstab(
    df_cross['EdLevel_Clean'], 
    df_cross['Employment'], 
    normalize='index'
) * 100

# Display the percentage breakdown table
print("Employment Type Breakdown by Education Level (%):")
print(ed_emp_crosstab.round(1))
```

```python
# Sort the breakdown to make the visual chart flow logically
ed_emp_crosstab = ed_emp_crosstab.sort_values(by='Employed, full-time', ascending=False)

# Plot the 100% Stacked Bar Chart
ax = ed_emp_crosstab.plot(kind='barh', stacked=True, figsize=(12, 7), cmap='viridis')

# Customizing titles, legends, and axis formats
plt.title('Employment Status Breakdown Across Education Levels', fontsize=14, pad=15)
plt.xlabel('Percentage of Respondents (%)', fontsize=12)
plt.ylabel('Educational Background (EdLevel)', fontsize=12)
plt.legend(title='Employment Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xlim(0, 100)

# Optional: Add percentage text markers inside the bars for readability
for p in ax.patches:
    width = p.get_width()
    if width > 5: # Only show labels for segments large enough to fit text
        x, y = p.get_xy()
        ax.text(x + width/2, y + p.get_height()/2, f'{width:.1f}%', 
                va='center', ha='center', color='white', fontsize=9, weight='bold')

plt.tight_layout()
plt.show()
```
