## Tasks


#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Create a comprehensive summary DataFrame
summary_df = pd.DataFrame({
    'Data Type': df.dtypes,
    'Non-Null Count': df.count(),
    'Missing Values': df.isnull().sum()
})

# Display the summary table
print(summary_df)
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Generate descriptive statistics for all numerical columns
numerical_summary = df.describe()

# Display the summary table
print(numerical_summary)
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# View all unique countries and how many times they appear
country_counts = df['Country'].value_counts()

# Print the entire series (or use pd.set_option('display.max_rows', None) to see all)
print(country_counts)
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
# Step 1: Strip accidental whitespace and enforce Title Case formatting
df['Country'] = df['Country'].str.strip().str.title()

# Step 2: Define a explicit mapping dictionary for identified inconsistencies
country_mapping = {
    'Usa': 'United States',
    'Us': 'United States',
    'United States Of America': 'United States',
    'Uk': 'United Kingdom',
    'United Kingdomm': 'United Kingdom',
    'Russian Federation': 'Russia'
}

# Step 3: Apply the mapping to standardize the values
df['Country'] = df['Country'].replace(country_mapping)
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
# Create one-hot encoded columns with a clean prefix
employment_dummies = pd.get_dummies(df['Employment'], prefix='Employment', dtype=int)

# Concat the new dummy columns back to the main DataFrame
df = pd.concat([df, employment_dummies], axis=1)

# Optional: Drop the original categorical Employment column if it's no longer needed
df.drop('Employment', axis=1, inplace=True)

# Preview the changes
print(df.filter(like='Employment').head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
# Calculate missing values per column and sort them from highest to lowest
missing_values_summary = df.isnull().sum().sort_values(ascending=False)

# Display the top 10 columns with the most missing values
print("Columns with the most missing values:")
print(missing_values_summary.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
# 1. Calculate the median of the column
median_value = df['ConvertedCompYearly'].median()
print(f"Median value used for imputation: {median_value}")

# 2. Fill missing values in place
df['ConvertedCompYearly'].fillna(median_value, inplace=True)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
# 1. Find the most frequent categorical value (the mode)
most_frequent_value = df['RemoteWork'].value_counts().idxmax()
print(f"Most frequent category (Mode): '{most_frequent_value}'")

# 2. Fill the missing values with that category in place
df['RemoteWork'].fillna(most_frequent_value, inplace=True)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
# 1. Calculate min and max values
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

# 2. Apply the Min-Max formula to create a new normalized column
df['ConvertedCompYearly_Scaled'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)

# 3. Verify the new range is exactly 0.0 to 1.0
print("Minimum scaled value:", df['ConvertedCompYearly_Scaled'].min())
print("Maximum scaled value:", df['ConvertedCompYearly_Scaled'].max())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
# Apply log transformation (Natural Log + 1)
df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

# Preview the original vs. transformed values
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_Log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
# Step 1: Clean 'YearsCodePro' text entries to numeric values
df['YearsCodePro'] = df['YearsCodePro'].replace('Less than 1 year', 0)
df['YearsCodePro'] = df['YearsCodePro'].replace('More than 50 years', 51)

# Convert the column to numeric data type, turning any unexpected strings/errors into NaN
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Optional: Fill missing experience values with the median before mapping
df['YearsCodePro'].fillna(df['YearsCodePro'].median(), inplace=True)

# Step 2: Define the categorization logic
def categorize_experience(years):
    if years < 2:
        return 'Junior'
    elif 2 <= years <= 5:
        return 'Intermediate'
    elif 5 < years <= 10:
        return 'Senior'
    else:
        return 'Executive/Lead'

# Step 3: Apply the function to create the new column
df['ExperienceLevel'] = df['YearsCodePro'].apply(categorize_experience)

# Step 4: Verify the distribution of the new column
print(df['ExperienceLevel'].value_counts())
```
